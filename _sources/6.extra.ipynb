```python
import pandas as pd
from IPython.display import display
from unidecode import unidecode
```

```python
df = pd.read_excel('datos_CAIMEDv8_unido_v5.xlsx')
```

```python
meds = ['alirocumab', 'Acido fenofibrico', 'Atorvastatina', 'Colestiramina', 'Evolocumab',
        'Ezetimibe', 'Fenofibrato'] 
```

```python
sum = 0
for med in meds:
    sum += (len(df[df[med]=='si']))
print(sum)
```

```python
hipo = ['no']*len(df)
for index in range(len(df)):
    line = df.iloc[index]
    
    for med in meds:
        if unidecode(line[med].lower())=='si':
            hipo[index] = 'si'
```

```python
hipo.count('si')
```

```python
sum = 0
for med in meds:
    print(med)
    sum += list(df[med]).count('si')
print(sum)
```
