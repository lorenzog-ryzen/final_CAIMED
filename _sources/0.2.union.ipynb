# Unión de Datos Nuevos

```python
import pandas as pd
from IPython.display import display
import numpy as np
df_0 = pd.read_excel('datos_CAIMEDv8_fixed.xlsx')
df_1 = pd.read_excel('Datos riesgo cardiovascular estudio CARED  para unirv5.xlsx')
```

```python
df_0['individuo'] = df_0['individuo'].astype(int)
df_1['cedula'] = df_1['cedula'].astype(int)
```

```python
df_1
```

```python
i = 0
for ced in df_1['cedula']:
    if ced not in list(df_0['individuo'].astype(int)):
        i += 1
        print('eroor', ced)

print(i, 'en total')
i = 0

for ced in df_0['individuo']:
    if ced not in list(df_1['cedula'].astype(int)):
        i += 1
        print('eroor', ced)
print(i, 'en total')
```

```python
column_names = list(df_1.columns)
```

```python
dx= {}
for name in column_names:
    dx[name] = []
for row_a in range(len(df_0)):
    founded = False
    index_founded = None
    current_ced  = (df_0.iloc[row_a])['individuo']
    if current_ced in list(df_1['cedula']):
        for column in column_names:
            if column == 'cedula' :
                dx[column].append(int(str(current_ced).replace('.','')))
            else:
                dx[column].append(  list(df_1[df_1['cedula']==current_ced][column])[0]     )
    else:
        for column in column_names:
            dx[column].append('')
```

```python
dx = pd.DataFrame(dx)
```

```python
dx['sexo']
```

```python
len(dx) == len(df_0)
```

```python
dx_2= {}
i = 0
aaa = {0:'a',
       1:'b'}
for name in (list(dx.columns) + list(df_0.columns)):
    if 'sexo' == name:
        dx_2[f'{name}_{aaa[i]}'] = []
        i+= 1
    dx_2[name] = []
for index in range(len(df_0)):
    for name in list(df_0.columns):
        if name == 'sexo':
            dx_2['sexo_a'].append((df_0.iloc[index])['sexo'])
        else:
            dx_2[name].append((df_0.iloc[index])[name])
    for name2 in dx.columns:
        if name2 == 'sexo':
            dx_2['sexo_b'].append((dx.iloc[index])['sexo'])
        else:
            dx_2[name2].append((dx.iloc[index])[name2])
```

```python
dx_2['sexo'] = dx_2['sexo_a']
```

```python
dx2 = pd.DataFrame(dx_2)
```

```python
# Ensure 'cedula' is of type string
dx2['cedula'] = dx2['cedula'].astype(str)

# Filter rows where 'cedula' is not empty
dx2 = dx2[dx2['cedula'] != '']

# Reset the index if required
dx2.reset_index(drop=True, inplace=True)
    
```
