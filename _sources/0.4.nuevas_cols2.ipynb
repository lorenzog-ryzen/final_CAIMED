# Modificaciones con Riesgo CV

- **Triglicéridos** <150 mg/dl y en los que presenten valores superiores se niveles más altos hacen necesario evaluar otros factores de riesgo.

  **Comentario:** Faltaría definir cuáles factores de riesgo, proponemos:
  - IMC >=30
  - Antecedente de hipertensión arterial o diabetes mellitus (DM)
  - LDL >70

---

- **C-LDL** Comentado en la tabla anterior
  - **Muy alto riesgo:** Un régimen terapéutico que logre >50% de reducción de C-LDL de su basal y un C-LDL <55 mg/dl.
  - **Alto riesgo:** Un régimen terapéutico que logre >50% de reducción de C-LDL basal y un C-LDL <70 mg/dl.
  - **Riesgo moderado:** Objetivo <100 mg/dl.
  - **Bajo riesgo:** Objetivo <116 mg/dl.

- **Colesterol no HDL** Se establecería como la diferencia entre valores reportados de Colesterol total – Colesterol HDL en la base de datos.
  - **Muy alto riesgo:** >85 mg/dl.
  - **Alto riesgo:** >100 mg/dl.
  - **Riesgo moderado:** >130 mg/dl.
  - **Bajo riesgo:** No tiene establecida meta.


```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('datos_CAIMEDv8_unido_v5.xlsx').drop( ['Muy_Alto_riesgo_cLDL_controlado',
       'Alto_riesgo_cLDL_controlado',
       'Muy_Alto_riesgo_HDL_controlado',
       'Alto_riesgo_HDL_controlado',
       'Trigliceridos_controlados'],axis='columns')
```

```python
for i in df:
    if 'HD' in i:
        print(i)
```

```python
list_names = ['Muy_Alto_riesgo_cLDL_controlado',
       'Alto_riesgo_cLDL_controlado',
       'Muy_Alto_riesgo_HDL_controlado',
       'Alto_riesgo_HDL_controlado',
       'Trigliceridos_controlados']
```

```python
list_names = ['Riesgo_cLDL_controlado',
       'Riesgo_HDL_controlado',
       'Trigliceridos_controlados']
```

```python
vect1 = ['no']*len(df)
vect2 = ['no']*len(df)
vect3 = ['no']*len(df)


for i in range(len(df)):
    cldl = (df.iloc[i])['cLDL(mg/dl)']
    cv =  (df.iloc[i])['Riesgo CV']
    hdl = (df.iloc[i])['HDL(mg/dl)']
    if cv == 'Muy alto':
        if cldl <55:
            vect1[i] = 'si'
        if hdl < 85:
            vect2[i] = 'si'
    if cv == 'Alto':
        if cldl <70:
            vect1[i] = 'si'
        if hdl < 100:
            vect2[i] = 'si'
    trigli = (df.iloc[i])['trigliceridos(mg/dl)']
    if trigli <=150:
        vect3[i] = 'si'
        
    
```

```python
dx = {'cLDL_controlado':vect1,
       'HDL_controlado':vect2,
       'Trigliceridos_controlados':vect3}
dx = pd.DataFrame(dx)
```
