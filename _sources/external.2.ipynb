```python
import pandas as pd
import numpy as np
from IPython.display import display
import datos
```

```python
for i in datos.datos :
    if 'tensi' in i:
        print(i)        
```

```python
hb1ac_control   = ['no controlado']*len(datos.datos) 
peso_control    = ['no controlado']*len(datos.datos) 
presion_control = ['no controlado']*len(datos.datos) 
perimtero_control = ['no controlado']*len(datos.datos) 
for row_index in range(len(datos.datos)):
    row = datos.datos.iloc[row_index]
    diab = row['Diabetico']
    hb1ac = row['POC_hba1c(%)']
    normo = row['normo_peso']
    per = row['perimetro_abdominal_normal']
    tens_0 = row['tension_normal'      ]
    tens_1 = row['tension_elevada'     ]
    tens_2 = row['tension_alta_nivel_1']
    
    if diab == 'No' and hb1ac <5.5:
        hb1ac_control[row_index] = 'controlado'
    elif diab == 'Si' and hb1ac <7: 
        hb1ac_control[row_index] = 'controlado'
    if normo == 'si':
        peso_control[row_index]  = 'controlado'
    if per == 'si':
        perimtero_control[row_index] = 'controlado'
    if tens_0 == 'si' or tens_1 == 'si' or tens_2 == 'si' :
        presion_control[row_index] = 'controlado'
    
    
        
    
    
```

```python
new_df = pd.DataFrame({
    'hb1ac_controlado':hb1ac_control,
    'peso_controlado':peso_control,
    'tension_controlada':presion_control,
    'perimtero_abdominal_controlado':perimtero_control
})
```

```python
final_df = pd.concat([datos.datos, new_df],axis=1)
```

```python
final_df.to_excel('datos_CAIMEDvX.xlsx')
```
