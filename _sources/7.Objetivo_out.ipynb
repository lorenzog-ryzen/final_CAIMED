```python
for index, var in enumerate(['Trigliceridos_controlados',
'hb1ac_controlado',
'peso_controlado',
'tension_controlada',
'perimtero_abdominal_controlado']):
    frequencies = df[var].value_counts()  # Conteo absoluto
    percentages = df[var].value_counts(normalize=True) * 100  
    freq_table = pd.DataFrame({
        "Count": frequencies,
        "Percentage (%)": percentages
    })

    print(freq_table)
```

```python
import pandas as pd

df_3 = pd.DataFrame()
var_central = 'colesterol_total_controlado'
for index, var in enumerate(['sexo', 'estrato_socieconomico', 'antecedentes_de_tabaquismo', 'ACV', 'escolaridad']):
    df_1 = pd.DataFrame()
    df_2 = pd.DataFrame()
    for posicion in ['controlado', 'no controlado']:
        if posicion == 'si':
            frequencies = df[df[var_central] == posicion][var].value_counts()  
            percentages = (df[df[var_central] == posicion][var].value_counts(normalize=True) * 100).round(2)  
            freq_table = pd.DataFrame({
                "Count": frequencies,
                "Percentage (%)": percentages
            })
            df_1 = pd.concat([df_1, freq_table], axis=1)    
        else:
            frequencies = df[df[var_central] == posicion][var].value_counts()  
            percentages = (df[df[var_central] == posicion][var].value_counts(normalize=True) * 100).round(2)  
            freq_table = pd.DataFrame({
                "Count": frequencies,
                "Percentage (%)": percentages
            })
            df_2 = pd.concat([df_2, freq_table], axis=1)

    temp_df = pd.concat([df_1, df_2], axis=1)
    temp_df = temp_df.sort_index(axis=0)
    df_3 = pd.concat([df_3, temp_df], axis=0) 

print(df_3)

```
