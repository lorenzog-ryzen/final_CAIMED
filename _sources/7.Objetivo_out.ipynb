```python
import pandas as pd
from scipy.stats import chi2_contingency

df_3 = pd.DataFrame()
var_central = 'colesterol_total_controlado'
for index, var in enumerate(['sexo', 'edad_grupo', 'estrato_socieconomico', 'antecedentes_de_tabaquismo', 'ACV', 'escolaridad']):
    df_1 = pd.DataFrame()
    df_2 = pd.DataFrame()
    for posicion in ['controlado', 'no controlado']:
        if posicion == 'si':
            frequencies = df[df[var_central] == posicion][var].value_counts()  
            percentages = (df[df[var_central] == posicion][var].value_counts(normalize=True) * 100).round(2)  
            freq_table = pd.DataFrame({"Count": frequencies,"Percentage (%)": percentages})
            df_1 = pd.concat([df_1, freq_table], axis=1)    
        else:
            frequencies = df[df[var_central] == posicion][var].value_counts()  
            percentages = (df[df[var_central] == posicion][var].value_counts(normalize=True) * 100).round(2)  
            freq_table = pd.DataFrame({"Count": frequencies,"Percentage (%)": percentages})
            df_2 = pd.concat([df_2, freq_table], axis=1)

    temp_df = pd.concat([df_1, df_2], axis=1)
    temp_df = temp_df.sort_index(axis=0)
    df_3 = pd.concat([df_3, temp_df], axis=0) 

with open('archive.txt','a') as file :
    file.write((df_3.to_string())[1:]) 
    
    file.close()
lenghts = [2, 4, 6, 2, 2, 12]
vars_names = ['sexo', 'edad_grupo', 'estrato_socieconomico', 'antecedentes_de_tabaquismo', 'ACV', 'escolaridad','Riesgo CV']

with open('archive.txt', 'r') as reading, open('0.final_archive.txt', 'a') as file_1:
    # Leer líneas del archivo excluyendo la cabecera
    lines = reading.readlines()[1:]
    print(lines[0])  # Verificar que las líneas se están leyendo correctamente
    
    index = 0
    for lenght, var_name in zip(lenghts, vars_names):
        # Generar la tabla de contingencia
        contingency_table = pd.crosstab(df[var_name], df[var_central])
        
        # Calcular Chi-cuadrado
        chi2, p, dof, _ = chi2_contingency(contingency_table)
        
        # Calcular la posición de espaciado dinámico
        posicion = 74 - len(var_name)
        
        # Escribir los resultados del Chi-cuadrado
        file_1.writelines(f'{var_name}{" " * posicion}{p:.4f}\n')
        
        # Seleccionar y escribir las líneas correspondientes
        selected_lines = lines[index:index + lenght]
        for line in selected_lines:
            file_1.writelines(f'{line.strip()}\n')
        
        # Actualizar el índice
        index += lenght

```

```python
import pandas as pd
from scipy.stats import chi2_contingency

df_3 = pd.DataFrame()
var_central = 'Trigliceridos_controlados'

for index, var in enumerate(['sexo', 'edad_grupo', 'estrato_socieconomico', 'antecedentes_de_tabaquismo', 'ACV', 'escolaridad','Riesgo CV','tipo_de_afiliacion']):
    df_1 = pd.DataFrame()
    df_2 = pd.DataFrame()

    # Calcular frecuencias y porcentajes por fila
    contingency_table = pd.crosstab(df[var], df[var_central], margins=False)
    percentages_by_row = (contingency_table.div(contingency_table.sum(axis=1), axis=0) * 100).round(2)

    # Combinar frecuencias y porcentajes
    freq_table = contingency_table.copy()
    for col in contingency_table.columns:
        freq_table[col] = contingency_table[col].astype(str) + " (" + percentages_by_row[col].astype(str) + "%)"

    df_3 = pd.concat([df_3, freq_table], axis=0)

# Guardar los resultados en un archivo
with open('archive.txt', 'w') as file:
    file.write(df_3.to_string())
    file.close()

# Continuar con el análisis de Chi-cuadrado y resultados
lenghts = [2, 4, 6, 2, 2, 12, 3,6]
vars_names = ['sexo', 'edad_grupo', 'estrato_socieconomico', 'antecedentes_de_tabaquismo', 'ACV', 'escolaridad', 'Riesgo CV','tipo_de_afiliacion']

with open('archive.txt', 'r') as reading, open('0.final_archive.txt', 'a') as file_1:
    # Leer líneas del archivo excluyendo la cabecera
    lines = reading.readlines()[1:]
    print(lines[0])  # Verificar que las líneas se están leyendo correctamente

    index = 0
    for lenght, var_name in zip(lenghts, vars_names):
        # Generar la tabla de contingencia
        contingency_table = pd.crosstab(df[var_name], df[var_central])

        # Calcular Chi-cuadrado
        chi2, p, dof, _ = chi2_contingency(contingency_table)
        posicion = 74 - len(var_name)
        file_1.writelines(f'{var_name}{" " * posicion}{p:.4f}\n')
        selected_lines = lines[index:index + lenght]
        for line in selected_lines:
            file_1.writelines(f'{line.strip()}\n')
        index += lenght
```
