```python
columns = [
    'no_diabeticos_POC_hba1c(%)_controlado',
    'diabeticos_POC_hba1c(%)_controlado', 
    'bajo_peso',
    'normo_peso',
    'sobrepeso',
    'obesidad',
    'perimetro_abdominal_normal',
    'perimetro_abdominal_aumentado',
    'perimetro_abdominal_incremento_sustancial',
    'tension_normal',
    'tension_elevada',
    'tension_alta_nivel_1',
    'tension_alta_nivel_2',
    'tension_crisis_hipertensiva',
]
```

```python
vect1 = ['no'] * len(datos.datos)
vect2 = ['no'] * len(datos.datos)
vect3 = ['no'] * len(datos.datos)
vect4 = ['no'] * len(datos.datos)
vect5 = ['no'] * len(datos.datos)
vect6 = ['no'] * len(datos.datos)
vect7 = ['no'] * len(datos.datos)
vect8 = ['no'] * len(datos.datos)
vect9 = ['no'] * len(datos.datos)
vect10 = ['no'] * len(datos.datos)
vect11 = ['no'] * len(datos.datos)
vect12 = ['no'] * len(datos.datos)
vect13 = ['no'] * len(datos.datos)
vect14 = ['no'] * len(datos.datos)

for row in range(len(datos.datos)):
    diab = datos.datos.iloc[row]['diabetes mellitus']
    hba = datos.datos.iloc[row]['POC_hba1c(%)']
    if diab == 'no':
        if hba < 5.7:
            vect1[row] = 'si'
    else:
        if hba < 7:
            vect2[row] = 'si'
    imc = datos.datos.iloc[row]['IMC']
    if imc < 19.9:
        vect3[row] = 'si'
    elif 20 <= imc < 25:
        vect4[row] = 'si'
    elif 25 < imc < 30:
        vect5[row] = 'si'
    else:
        vect6[row] = 'si'
    sexo = datos.datos.iloc[row]['sexo']
    pr_abd = datos.datos.iloc[row]['perimetro_abdominal']
    if sexo == 'masculino':
        if pr_abd < 94:
            vect7[row] = 'si'
        elif pr_abd < 102:
            vect8[row] = 'si'
        else:
            vect9[row] = 'si'
    else:
        if pr_abd < 80:
            vect7[row] = 'si'
        elif pr_abd < 88:
            vect8[row] = 'si'
        else:
            vect9[row] = 'si'
    sistolic = datos.datos.iloc[row]['presion_arterial_sistolica']
    diastolic = datos.datos.iloc[row]['presion_arterial_diastolica']
    if sistolic < 120 and diastolic < 80:
        vect10[row] = 'si'
    elif 120 <= sistolic <= 129 and diastolic < 80:
        vect11[row] = 'si'
    elif 130 <= sistolic <= 139 or 80 <= diastolic <= 89:
        vect12[row] = 'si'
    elif sistolic >= 140 or diastolic >= 90:
        vect13[row] = 'si'
    elif sistolic > 180 or diastolic > 120:
        vect14[row] = 'si'

datos.datos['no_diabeticos_POC_hba1c(%)_controlado'] = vect1
datos.datos['diabeticos_POC_hba1c(%)_controlado'] = vect2
datos.datos['bajo_peso'] = vect3
datos.datos['normo_peso'] = vect4
datos.datos['sobrepeso'] = vect5
datos.datos['obesidad'] = vect6
datos.datos['perimetro_abdominal_normal'] = vect7
datos.datos['perimetro_abdominal_aumentado'] = vect8
datos.datos['perimetro_abdominal_incremento_sustancial'] = vect9
datos.datos['tension_normal'] = vect10
datos.datos['tension_elevada'] = vect11
datos.datos['tension_alta_nivel_1'] = vect12
datos.datos['tension_alta_nivel_2'] = vect13
datos.datos['tension_crisis_hipertensiva'] = vect14
```
