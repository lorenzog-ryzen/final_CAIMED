# Rovustina cambios

```python
import pandas as pd
import numpy  as np
from IPython.display  import display
```

```python
df_0  =  pd.read_excel('datos_CAIMEDvXI_Rosuvastina.xlsx')
df_0  =  df_0[['individuo','Rosuvastina','Insulina','Antidiabético', 'Hipolipemiante']]
df_1  =  pd.read_excel('datos_CAIMEDvXII_trigliceridos.xlsx')
```

```python
vcts = [['no']*len(df_1),
        ['no']*len(df_1),
        ['no']*len(df_1),
        ['no']*len(df_1)]
for index, ced in enumerate( df_1['cedula']):
    if ced in list(df_0['individuo']):
        row = df_0[df_0['individuo']==ced]
        vcts[0][index] = (row)['Rosuvastina']
        vcts[1][index] = (row)['Insulina']
        vcts[2][index] = (row)['Antidiabético']
        vcts[3][index] = (row)['Hipolipemiante']
```

```python
df_1['Rosuvastina'] = vcts[0]
df_1['Insulina']    = vcts[1]
df_1['Antidiabético'] = vcts[2]
df_1['Hipolipemiante'] = vcts[3]
```

Las columnas mencionadas fueron actualizadas para garantizar un manejo adecuado de la información que contienen.
